```python
import pandas as pd
```

```python
df = pd.read_csv('Offtake.csv')
```

```python
# Assign financial year based on FY ending year
df["FinancialYear"] = df["Year"] + (df["Month"] >= 7).astype(int)

# Calculate financial year totals for columns 1-5
fy_totals = (
    df.groupby(["Name", "FinancialYear"])[["1", "2", "3", "4", "5"]]
      .sum()
      .reset_index()
)

print(fy_totals)
```
